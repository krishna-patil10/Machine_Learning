```python
import seaborn as sns
import pandas as pd
```

```python
dict1 = {
    'x1': [i for i in range(1, 10)],
    'x2': [i for i in range(11, 20)],
    'x3': [i for i in range(30, 21, -1)],
    'x4': [i for i in range(41, 50)],
    'y': [i for i in range(101, 110)]
}
```

```python
df = pd.DataFrame(data=dict1)
df
```

```python
df.corr()
```

```python
df1 = sns.load_dataset('tips')[['total_bill', 'tip', 'size']]
df1
```

```python
df1.corr()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(4, 4))
sns.heatmap(df1.corr(), annot=True, cmap='coolwarm')
plt.show()
```

```python
df1.head()
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
ss = StandardScaler()
min_max = MinMaxScaler()
rs = RobustScaler()
```

```python
scaled_size = ss.fit_transform(df1[['size']])
scaled_tip = min_max.fit_transform(df1[['tip']])
scaled_total_bill = rs.fit_transform(df1[['total_bill']])
```
