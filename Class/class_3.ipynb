```python
import pandas as pd
import seaborn as sns

df= sns.load_dataset('tips')
```

```python
from sklearn.preprocessing import LabelEncoder
encoder_time=LabelEncoder()

encoded_data= encoder_time.fit_transform(df['time'])
```

```python
print('classes',encoder_time.classes_)
```

```python
encoded_data
```

```python
for category, number in zip (encoder_time.classes_, range(len(encoder_time.classes_))):
    print(category, '-------',number)
```

```python
encoder_day=LabelEncoder()

encoded_day_data= encoder_day.fit_transform(df['day'])
```

```python
print('classes',encoder_day.classes_)
```

```python
for category, number in zip (encoder_day.classes_,range(len(encoder_day.classes_))):
    print(category,'----------',number)
```

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer
```

```python
# A right-skewed feature
x=np.array([[-1],[0],[3],[5],[10],[50]])

#create the transformer
pt=PowerTransformer(method='yeo-johnson')

#fit and transform
x_transformed= pt.fit_transform(x)

print('original:')
print(x.ravel())


print('\nTransformed:')
print(np.round(x_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
# A right-skewed feature
y=np.array([[1],[2],[3],[5],[10],[50]])

#create the transformer
pt=PowerTransformer(method='box-cox')

#fit and transform
y_transformed= pt.fit_transform(y)

print('original:')
print(y.ravel())


print('\nTransformed:')
print(np.round(y_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
import numpy as np
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry'],
    'Gender': ['F', 'M', None, 'M', 'F', 'M', None, 'M'],                     # Categorical with Nulls
    'City': ['Mumbai', 'Delhi', 'Bangalore', None, 'Mumbai', None, 'Pune', 'Delhi'], # Categorical with Nulls
    'Department': ['IT', 'HR', None, 'Finance', 'IT', 'Marketing', 'HR', None],  # Categorical with Nulls
    'Age': [25, np.nan, 30, 35, np.nan, 40, 28, 33],                           # Numerical with Nulls
    'Experience': [2, 4, np.nan, 10, 3, np.nan, 5, 8],                          # Numerical with Nulls
    'Salary': [50000, 62000, 75000, np.nan, 58000, 90000, np.nan, 82000]       # Numerical with Nulls
}

df = pd.DataFrame(data)
df
```

```python
df.head()
```

```python
from sklearn .impute import SimpleImputer
```

```python
imputer = SimpleImputer(strategy='mean')

imputer.fit_transform(df[['Salary']])
```

```python
imputer = SimpleImputer(strategy='mean')

df['new_salary']=imputer.fit_transform(df[['Salary']]).flatten()
```

```python
imputer = SimpleImputer(strategy='median')

df['experience']=imputer.fit_transform(df[['Experience']]).flatten()
```

```python
imputer = SimpleImputer(strategy='most_frequent',)

df['City']=imputer.fit_transform(df[['City']]).flatten()
```
