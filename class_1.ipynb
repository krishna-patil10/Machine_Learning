```python
def variance_std (*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_variance= total/(len(args))
    return calculated_variance , calculated_variance**(1/2)
```

```python
variance_std(*array)
```

```python
def variance_std (*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_variance= total/(len(args)-1)
    return calculated_variance , calculated_variance**(1/2)
```

```python
variance_std(*array)
```

```python
hours=[1,2,3,4,5]
marks=[30,40,45,50,65]

import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(3,3))
plt.scatter(hours,marks,marker='+',color='r')
plt.xlabel('no of hours')
plt.ylabel('marks')
plt.title('hours vs marks')
plt.show()
```
