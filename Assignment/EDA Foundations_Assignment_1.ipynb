5.  Develop a complete data-preparation pipeline in Python that creates the DataFrame, validates the feature and target columns, generates the train-test split, and prints a concise report of the resulting datasets. Explain two potential mistakes that could make the pipeline unreliable.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Step 1: Create the DataFrame
data = {
    'Student_ID': [101, 102, 103, 104, 105, 106,
                   107, 108, 109, 110, 111, 112],
    'Name': ['Amit', 'Priya', 'Rahul', 'Neha', 'Vikas', 'Pooja',
             'Karan', 'Simran', 'Arjun', 'Meena', 'Rohit', 'Anjali'],
    'Age': [19, 20, 18, 21, 20, 19, 22, 18, 21, 20, 19, 22],
    'Hours_Studied': [2, 5, 3, 8, 6, 4, 9, 2, 7, 5, 6, 8],
    'Attendance': [65, 85, 70, 95, 88, 75, 98, 60, 90, 82, 86, 94],
    'Marks': [48, 78, 55, 92, 84, 67, 96, 45, 89, 76, 81, 93]
}

df = pd.DataFrame(data)

# Step 2: Define features and target
feature_columns = ['Hours_Studied', 'Attendance']
target_column = 'Marks'

# Step 3: Validate required columns
required_columns = feature_columns + [target_column]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")

# Check for missing values
if df[required_columns].isnull().any().any():
    raise ValueError("Missing values found in features or target.")

# Check that feature and target values are numeric
if not all(
    pd.api.types.is_numeric_dtype(df[col])
    for col in required_columns
):
    raise TypeError("Features and target must be numeric.")

# Step 4: Create X and y
X = df[feature_columns]
y = df[target_column]

# Step 5: Generate reproducible train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Step 6: Print a concise report
print("===== DATA PREPARATION REPORT =====")
print("Original DataFrame:", df.shape)
print("Features:", feature_columns)
print("Target:", target_column)

print("\nTraining features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)

print("\nTraining dimensions consistent:",
      len(X_train) == len(y_train))

print("Testing dimensions consistent:",
      len(X_test) == len(y_test))

print("Index overlap:",
      not set(X_train.index).isdisjoint(set(X_test.index)))

print("\nPipeline completed successfully!")
```
